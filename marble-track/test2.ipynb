```python
import pygame
from pygame import Color

# Initialize Pygame
pygame.init()

# Set screen dimensions
width, height = 800, 600
screen = pygame.display.set_mode((width, height))
pygame.display.set_caption("Marble Track Simulation")

# Colors
white = Color("red")
black = (0, 0, 0)
blue = (0, 0, 255)

# Marble properties
marble_radius = 15
marble_x, marble_y = 50, 50
marble_speed_x, marble_speed_y = 2, 0
gravity = 0.1

# Track points (x, y)
track_points = [(50, 100), (200, 50), (350, 150), (500, 100), (700, 150)]

# Game loop
running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    # Clear the screen
    screen.fill(white)

    # Draw the track
    pygame.draw.lines(screen, black, False, track_points, 5)

    # Update marble position
    marble_x += marble_speed_x
    marble_y += marble_speed_y
    marble_speed_y += gravity

    # Simple collision detection (very basic)
    for point in track_points:
        if (marble_x - point[0])**2 + (marble_y - point[1])**2 <= (marble_radius + 5)**2:
            marble_speed_y *= -0.5  # Reverse and dampen vertical speed
            marble_y = point[1] - marble_radius - 5 #adjust position

    # Keep marble within screen bounds
    if marble_x + marble_radius > width or marble_x - marble_radius < 0:
        marble_speed_x *= -1
    if marble_y + marble_radius > height:
        marble_speed_y *= -0.8
        marble_y = height - marble_radius
    if marble_y - marble_radius < 0:
        marble_speed_y *= -0.8
        marble_y = marble_radius

    # Draw the marble
    pygame.draw.circle(screen, blue, (marble_x, marble_y), marble_radius)

    # Update the display
    pygame.display.flip()

# Quit Pygame
pygame.quit()
```
